Question 3

```python
def minmax(A,axis):
    min = A.min(axis=axis,keepdims=True)
    max = A.max(axis=axis,keepdims=True)
    return (A-min)/(max-min)
```

```python
A_0 = minmax(A,axis=0)
print(A_0.min(axis=0))
print(A_0.max(axis=0))  
```

```python
A_1 = minmax(A,axis=1)
print(A_1.min(axis=1))
print(A_1.max(axis=1))  
```

Question 5

```python
col_means = A.mean(axis=0)
col_stds = A.std(axis=0)
```

```python
outlier_mask = np.abs(A - col_means) > 3 * col_stds
```

```python
print(outlier_mask.sum(axis=0))
```

```python
col_medians = np.median(A, axis=0)
```

```python
_, outlier_cols = np.where(outlier_mask)
A[outlier_mask] = col_medians[outlier_cols]
A
```

Question 9

```python
from numpy.lib.stride_tricks import sliding_window_view
```

```python
data = np.random.randint(0, 100, size=10000)
data
```

```python
windows = sliding_window_view(data, 10)
print(windows.shape)
```

```python
windows_mean = windows.mean(axis=1)
windows_mean.shape
```

```python
import pandas as pd
pd_means = pd.Series(data).rolling(10).mean()
print(np.allclose(windows_mean, pd_means[9:].to_numpy()))
```
